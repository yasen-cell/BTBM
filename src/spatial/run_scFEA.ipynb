This example runs scFEA on spatial data using the complete human/mouse metabolic modules.

**Parameters:**

- `--data_dir`: Directory containing scFEA model files.
- `--input_dir`: Directory containing input data.
- `--test_file`: Input single cell data file (e.g., Human Striatum or Mouse SubNigra), where rows are genes and columns are samples.
- `--moduleGene_file`: `module_gene_m168.csv(for human) or module_gene_complete_mouse_m168.csv(for mouse)`; contains genes for each module.
- `--stoichiometry_matrix`: `cmMat_c70_m168.csv(for human) or cmMat_complete_mouse_c70_m168.csv(for mouse) `; describes the relationship between compounds and modules. Each row is an intermediate metabolite, and each column is a metabolic module.
- `--cName_file`: `cName_c70_m168.csv`; The name of compounds. The table contains two rows. First row is compounds name and second row is corresponding id.


For human data

```python
%%bash
cd /scFEA
python src/scFEA.py --data_dir data --input_dir input \
                    --test_file single_cell.csv \             
                    --moduleGene_file module_gene_m168.csv \
                    --stoichiometry_matrix cmMat_c70_m168.csv \
                    --cName_file cName_c70_m168.csv
```

For mouse data

```python
%%bash
cd /scFEA
python src/scFEA.py --data_dir data --input_dir input \
                    --test_file single_cell.csv \
                    --moduleGene_file module_gene_complete_mouse_m168.csv \
                    --stoichiometry_matrix cmMat_complete_mouse_c70_m168.csv \
                    --sc_imputation True \
                    --cName_file cName_c70_m168.csv
```
